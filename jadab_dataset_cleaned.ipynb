```python
df[["ai_before_attempt"]] = OrdinalEncoder(categories=[["Never", "Rarely", "Sometimes", "Often", "Always"]]).fit_transform(df[["ai_before_attempt"]])
```

```python
df[["copy_paste_freq"]] = OrdinalEncoder(categories=[["Never", "Rarely", "Sometimes", "Often", "Always"]]).fit_transform(df[["copy_paste_freq"]])
```

```python
df[["answer_verification"]] = OrdinalEncoder(categories=[['Sometimes', 'Often', 'Always', 'Rarely', 'Never']]).fit_transform(df[["answer_verification"]])
```

```python
encoder = OrdinalEncoder(
    categories=[["Sometimes", "Daily", "Multiple times a day"]]
)

df[["ai_freq"]] = encoder.fit_transform(df[["ai_freq"]])
```

```python
encoder = OrdinalEncoder(
    categories=[["<1 hour", "1-3 hours", "3-5 hours", "5+ hours"]]
)
df[["ai_hours"]] = encoder.fit_transform(df[["ai_hours"]])
```

```python
df = pd.get_dummies(df, columns=["ai_purpose"], dtype=int)
```

```python
df["not_university"] = df["cgpa"].str.startswith("I am not").astype(int)
```

```python
df["cgpa"] = df["cgpa"].map({
    "< 3.00": 1,
    "3.00 - 3.5": 2,
    ">3.5": 3
})
```

```python
df["not_uni_sem"] = df["sem"].astype(str).eq(
    "Other (School or Collage or Higher Degree)"
).astype(int)
```

```python
df["sem"] = pd.to_numeric(df["sem"], errors="coerce")
```
